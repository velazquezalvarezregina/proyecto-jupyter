```
import matplotlib.pyplot as plt

# 1. HISTOGRAMA
valores = [10, 15, 12, 18, 20]
plt.bar(valores, valores, color='#4a90e2')
plt.title('Histograma')
plt.show()

# 2. DASHBOARD VENTAS
meses = ['Ene','Feb','Mar','Abr','May','Jun']
ventas = [1200,1800,1500,2400,2200,3000]
plt.bar(meses, ventas, color='#7ec8e3')
plt.title('Dashboard de Ventas')
plt.show()

# 3. PELICULAS
nombres = ['Interestelar','Matrix','Coco','Toy Story','Titanic','Avatar']
puntos = [8.7,8.7,8.4,8.3,7.9,7.8]
plt.barh(nombres, puntos, color='#a8e6cf')
plt.title('Peliculas Mejor Valoradas')
plt.show()

# 4. POPULARIDAD
vistas = [15,22,18,30,25,35]
plt.bar(nombres, vistas, color='#ffd93d')
plt.title('Popularidad')
plt.xticks(rotation=20)
plt.show()
```
